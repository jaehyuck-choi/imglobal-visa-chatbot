## 4. 지역 ↔ 관할청 매칭 (JOIN)

```python
%%sql
SELECT r.province AS 시도, r.city AS 시군구, o.name AS 관할청, ro.is_primary AS 대표여부
FROM region_offices ro
JOIN regions r ON r.id = ro.region_id
JOIN offices o ON o.id = ro.office_id
ORDER BY r.province, r.city, ro.is_primary DESC;
```

## 10. 예시: 강원특별자치도 철원군 담당 관할청

```python
%%sql
SELECT r.province, r.city, o.name, o.phone, o.address
FROM regions r
JOIN region_offices ro ON ro.region_id = r.id
JOIN offices o ON o.id = ro.office_id
WHERE r.province LIKE '강원%' AND r.city = '철원군';
```
